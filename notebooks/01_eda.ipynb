## Key Findings (EDA)

**Data quality**
- 284,807 rows originally, no missing values. 1,081 duplicate rows were removed.
- Class imbalance: about 473 fraud transactions out of about 283,726 (~0.17%). [verify with value_counts]
- Accuracy is a misleading metric here. I will use PR-AUC, precision and recall.

**Feature relationships (balanced sample, 473 fraud + 473 normal)**
- Strongest negative correlation with Class: V14 (-0.74), V12 (-0.68), V10 (-0.63).
- Strongest positive correlation with Class: V4 (0.70), V11 (0.69), V2 (0.47).
- Amount (0.045) and Time (-0.18) show little linear correlation. They may still help non-linear models, so they are not dropped.
- These correlations come from a balanced sample, so they overstate the real-world values. They are used only to rank features.

**Class separation (boxplots, full deduplicated data)**
- V14, V12, V10, V4 and V11 separate fraud from normal clearly: the boxes do not overlap.
- V2 separates weakly: the boxes are close and the whiskers overlap.
- Normal transactions cluster tightly around 0, while fraud is shifted and far more spread out.
- Outliers are kept, since extreme values can be the fraud signal.
- No single feature classifies perfectly, so a model combining all features is needed.

**Decisions for modeling**
- Split train/test from the full deduplicated data, stratified. Do not use the balanced sample for modeling.
- Scale Time and Amount (the only non-PCA columns).
- Handle imbalance with class weights and/or SMOTE (training data only).

```python
sns.histplot(x=np.log1p(dataframe['Amount']),kde=True)
```

**Amount distribution (log1p scale)**
- On the raw scale, Amount is extremely right-skewed (most values are small, a few reach ~$25,000), so log1p is used.
- After the transform there is a broad hump around ~$20 to $150 with sharp spikes at round prices ($1, $2, $10).
- The tallest spike is ~$1 (about 16,000 transactions), which suggests common fixed-price purchases.
- Amount will be log-transformed (or scaled) before modeling.

```python
sns.boxplot(y=np.log1p(dataframe['Amount']),x=dataframe['Class'])
```

**Amount by class (log1p scale)**
- Fraud median is ~$10 vs ~$21 for normal, and the fraud box is wider (~$1 to $110).
- The boxes overlap heavily, so Amount alone separates fraud weakly. This agrees with its low correlation (0.045).
- Normal has many high outliers (up to ~$25,000), while the largest fraud amount is ~$2,000.

```python
zero_amount_fraud=dataframe[(dataframe['Amount']==0 ) & (dataframe['Class']==1)].shape[0]
```

```python
zero_amount=dataframe[dataframe['Amount']==0].shape[0]
```

```python
zero_amount_fraud_per=zero_amount_fraud/zero_amount *100
zero_amount_fraud_per
```

**Zero-amount transactions**
- 25 of 473 fraud cases (~5%) have Amount = 0.
- Among all zero-amount rows, 1.3% are fraud, about 8x the overall fraud rate (~0.17%).
- Decision: keep these rows (dropping them would remove fraud examples) and test an `is_zero_amount` flag in feature engineering.
- Caution: this rests on only 25 fraud cases, so it is an elevated risk, not a rule.

```python
sns.histplot(x=dataframe['Time']/3600,kde=True,hue=dataframe['Class'],stat='density',common_norm=False)
```

**Time distribution (hours since first transaction)**
- Transactions follow a clear daily cycle: two humps about 24 hours apart, with low-activity periods around hours 0-7 and 26-31.
- Fraud (density, each class normalized separately) is flatter than normal and relatively more common during the low-activity hours, and relatively less common in the busiest stretches.
- Individual fraud spikes are noise (~10 fraud rows per hourly bin); only the overall trend is trusted.
- Time is measured from the first transaction, so hour 0 is not a known clock time. "Low-activity" is used instead of "night".
- Decision: test an `hour_of_day` feature (Time in hours modulo 24) against raw Time in feature engineering.
